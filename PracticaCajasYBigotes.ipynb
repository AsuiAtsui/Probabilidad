```python
import random
import matplotlib.pyplot as plt
from statistics import *
```

```python
fabrical = []
for i in range(100):
  numero_aleatorio = round(random.uniform(50,100),2)
  fabrical.append(numero_aleatorio)
print(fabrical)
```

```python
fabrica2 = []
for i in range(100):
  numero_aleatorio = round(random.uniform(40,90),2)
  fabrica2.append(numero_aleatorio)
print(fabrica2)
```

```python
plt.boxplot(fabrical)
plt.ylabel("Produccion en mts.")
plt.title("Produccion de la fabrica 1")
plt.show()
```

```python
plt.boxplot(fabrica2)
plt.ylabel("Produccion en mts.")
plt.title("Produccion de la fabrica 2")
plt.show()
```

```python
fig, (ax1,ax2) = plt.subplots(nrows=1, ncols=2,figsize=(9,4))
bplot1 = ax1.boxplot(fabrical,
                     vert=True, #vertical box alignment
                     patch_artist=True) #fill with color
ax1.set_title("Produccion de la fabrica 1")

bplot2 = ax2.boxplot(fabrica2,
                     vert=True, #vertical box alignment
                     patch_artist=True) #fill with color

ax2.set_title("Produccion de la fabrica 1")
plt.show()
```
