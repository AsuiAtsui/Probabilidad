```python
#Importar
from statistics import *

#Lista de calificaciones del grupo a de probabilidad y estadistica
grupoa = [70,70, 65,90,100,95,45,80,73,20,100,96]
print (grupoa)
type(grupoa)

#Sacar la media de calificaciones
promedio = sum(grupoa) / len(grupoa)
print(promedio)

#Sacar la media con la funcion mean
promedio2 = mean(grupoa)
print(promedio2)

#Determinar el valor de la mediana
mediana = median(grupoa)
print(mediana)

#Determinar la moda
moda = mode(grupoa)
print(moda)

#En caso de existir mas de una moda
moda2 = multimode(grupoa)
print(moda2)


#Tiempos que tardaron los 10 alumnos para correr los 100 metros

tiempos100mts = [15.10,17.20,14.69,13.27,22.15,18.71,19,18.9,20,21.41]
print("El tiempo promedio de los corredores es: ",mean(tiempos100mts))
print("El tiempo medio de los corredores es: ",median(tiempos100mts))
print("La moda de los corredores es: ",mode(tiempos100mts))
print("La multimoda de los corredores es: ",multimode(tiempos100mts))
```
